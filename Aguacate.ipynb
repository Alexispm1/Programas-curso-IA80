```python
import numpy as np
import statistics as stats
import matplotlib.pyplot as plt

# Pesos de los aguacates en gramos 🍏🍎
aguacates = [300, 500, 600, 400, 275, 700, 600]

# Estadísticas descriptivas 📊
media = np.mean(aguacates)  # Promedio
mediana = np.median(aguacates)  # Valor central
moda = stats.mode(aguacates)  # Valor más común
desviacion = np.std(aguacates)  # Variabilidad
# Coeficiente de Variación (CV): Mide la dispersión relativa. ¡Homogeneidad clave!
# Se expresa en porcentaje: (Desviación Estándar / Media) * 100
coeficiente_variacion = (desviacion / media) * 100

# Resultados 🍏📊
print(f"📊 Análisis de la cosecha de aguacates")
print(f"Media: {media}g")
print(f"Mediana: {mediana}g")
print(f"Moda: {moda}g")
print(f"Desviación estándar: {desviacion:.2f}g")
print(f"coeficiente_variacion: {coeficiente_variacion:.2f}g")

longitud = len(aguacates)
ideal = 0
menor = aguacates[0]
mayor = aguacates[0]
for i in range(longitud):
   if aguacates[i] < menor:
      menor = aguacates[i]
   else:
      mayor = aguacates[i]
   if aguacates[i] >= 400 and aguacates[i] <= 600:
     ideal = ideal + 1.

print (f"Ideales: {ideal} Menor: {menor} Mayor: {mayor}")

rangod = input("Cual es su rango DESDE ideal")
desde = int(rangod)
rangoh = input("Cual es su rango HASTA ideal")
hasta = int(rangoh)
if hasta < desde:
  print ("Ud. elijio un mal rango")
  exit
print (f"Ud. elijio desde: {desde} hasta: {hasta}")



# Crear el histograma
plt.hist(aguacates, bins=8, edgecolor='black')
# Etiquetas y título
plt.title("Histograma de distribución de pesos")
plt.xlabel("Peso (kg)")
plt.ylabel("Frecuencia")
# Mostrar el gráfico
plt.show()

# Crear el boxplot
plt.boxplot(aguacates, vert=False, patch_artist=True, boxprops=dict(facecolor='lightblue'))
# Etiquetas y título
plt.title("Boxplot de distribución de pesos")
plt.xlabel("Peso (kg)")
# Mostrar el gráfico
plt.show()

media = np.mean(aguacates)
desviacion = np.std(aguacates, ddof=1)  # ddof=1 → muestra
n = len(aguacates)

asimetria_manual = (n / ((n - 1) * (n - 2))) * np.sum(((aguacates - media) / desviacion) ** 3)
print(f"Asimetría (manual): {asimetria_manual:.4f}")

```
